5.1.2 The second step to use Pandas

```python
import numpy as np
np.random.seed(100)
a = np.random.standard_normal((9,4))
a
```

```python
import pandas as pd
df = pd.DataFrame(a)
df
```

```python
df.columns = ['No1','No2','No3','No4']
df
```

```python
df['No2'].mean()
```

```python
dates = pd.date_range('2019-1-1',periods=9,freq='ME')
dates
```

```python
df.index=dates
df
```
