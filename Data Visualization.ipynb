7.1.1 one-dimensional dataset

```python
import numpy as np
np.random.seed(1000)
y=np.random.standard_normal(20)
x = np.arange(len(y))
plt.plot(x, y)
plt.show()
```

```python
plt.plot(y)
```

```python
plt.plot(y.cumsum())
```

```python
plt.plot(y.cumsum())
plt.grid(False)
plt.axis('equal')
```

```python
plt.plot(y.cumsum())
plt.xlim(-1,20)
plt.ylim(np.min(y.cumsum())-1, np.max(y.cumsum())+1)
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.plot(y.cumsum(),'b',lw=1.5)
plt.plot(y.cumsum(),'ro')
plt.xlabel('index')
plt.ylabel('value')
plt.title('A simple plot')
    
```
