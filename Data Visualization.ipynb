7.1 static 2D plot

```python
import matplotlib.pyplot as plt

plt.style.use('seaborn-v0_8')
plt.rcParams['font.family'] = 'serif'
plt.rcParams['font.serif'] = ['Times New Roman']

%matplotlib inline
```

7.1.1 one-dimensional dataset

```python
import numpy as np
np.random.seed(1000)
y=np.random.standard_normal(20)
x = np.arange(len(y))
plt.plot(x, y)
plt.show()
```

```python
plt.plot(y)
```

```python
plt.plot(y.cumsum())
```

```python
plt.plot(y.cumsum())
plt.grid(False)
plt.axis('equal')
```

```python
plt.plot(y.cumsum())
plt.xlim(-1,20)
plt.ylim(np.min(y.cumsum())-1, np.max(y.cumsum())+1)
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.plot(y.cumsum(),'b',lw=1.5)
plt.plot(y.cumsum(),'ro')
plt.xlabel('index')
plt.ylabel('value')
plt.title('A simple plot')
    
```

7.1.2 one-dimensional dataset

```python
import matplotlib.pyplot as plt
import numpy as np
y = np.random.standard_normal((20,2)).cumsum(axis=0)
plt.figure(figsize=(10,6))
plt.plot(y,lw=1.5)
plt.plot(y,'ro')
plt.xlabel('index')
plt.ylabel('value')
plt.title('A simple plot')
```

```python
plt.figure(figsize=(10,6))
plt.plot(y[:,0],lw=1.5,label='y1')
plt.plot(y[:,1],lw=1.5,label='y2')
plt.plot(y,'ro')
plt.legend(loc=0)
plt.xlabel('index')
plt.ylabel('value')
plt.title('A simple plot')

```

```python
y[:,0]=y[:,0]*100
plt.plot(y[:,0],lw=1.5,label='y1')
plt.plot(y[:,1],lw=1.5,label='y2')
plt.plot(y,'ro')
plt.legend(loc=0)
plt.xlabel('index')
plt.ylabel('value')
plt.title('A simple plot')
```

```python
fig,axl=plt.subplots()
plt.plot(y[:,0],lw=1.5,label='y1')
plt.legend(loc=8)
plt.xlabel('index')
plt.ylabel('value 1st')
plt.title('A simple plot')
ax2=axl.twinx()
plt.plot(y[:,1],'g',lw=1.5,label='y2')
plt.plot(y[:,1],'ro')
plt.legend(loc=0)
plt.ylabel('value 2nd')

```

```python
plt.figure(figsize=(10, 6))
plt.subplot(211)
plt.plot(y[:,0],lw=1.5,label='1st')
plt.plot(y[:,0],'ro')
plt.legend(loc=0)
plt.ylabel('value')
plt.title('a simle plot')
plt.subplot(212)
plt.plot(y[:,1],'g',lw=1.5,label='2st')
plt.plot(y[:,1],'ro')
plt.legend(loc=0)
plt.ylabel('value')
plt.xlabel('index')

        
```

```python
plt.figure(figsize=(10, 6))
plt.subplot(121)
plt.plot(y[:,0],lw=1.5,label='1st')
plt.plot(y[:,0],'ro')
plt.legend(loc=0)
plt.ylabel('value')
plt.xlabel('index')
plt.title('1st data set')
plt.subplot(122)
plt.bar(np.arange(len(y)),y[:,1],width=0.5,color='g',label='2st')
plt.legend(loc=0)
plt.xlabel('index')
plt.title('2st data set')
```
