```python
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

np.random.seed(42)
data1 = np.random.normal(0, 1, 800)
data2 = np.random.normal(2, 1.2, 800)
data3 = np.random.normal(-1, 0.8, 800)
data_all = [data1, data2, data3]

plt.boxplot(data_all, patch_artist=True, widths=0.6,
            boxprops=dict(facecolor='#60a9ff', alpha=0.7),
            medianprops=dict(color='red'))
plt.xticks([1,2,3], ["Group 1", "Group 2", "Group3"])
plt.title('Multiple Boxplots')
plt.ylabel('Value')
plt.grid(axis='y',alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon
def func(x):
    return 0.5*np.exp(x)+1
a,b = 0.5,1.5
x = np.linspace(0,2, 100)
y = func(x)
Ix = np.linspace(a,b,100)
Iy = func(Ix)
verts = [(a,0)] + list(zip(Ix,Iy)) + [(b,0)]
fig, ax = plt.subplots(figsize=(10,6))
ax.plot(x,y,'b',linewidth=2)
ax.set_ylim(bottom=0)poly = Polygon(verts, facecolor='0.7', edgecolor='0.5')
poly.set_alpha(0.3)
ax.add_patch(poly)
ax.text(0.5*(a+b), 1, r'$\int_a^b f(x) \mathrm{d}x$', horizontalalignment='center', fontsize=20)
fig.text(0.9,0.075,'$x$')
fig.text(0.075,0.9,'$f(x)$')
ax.set_xticks([a, b]) 
ax.set_xticklabels(('$a$','$b$'))
ax.set_yticks([func(a), func(b)])
ax.set_yticklabels(('$f(a)$','$f(b)$'))
plt.show()
```

7.2  static 3D plot

```python
import numpy as np
strike = np.linspace(50, 150, 24)
ttm = np.linspace(0.5, 2.0, 24)
strike, ttm = np.meshgrid(strike, ttm)
strike[:,2].round(1)
```

```python
iv = (strike -100)**2/(strike*100)/ttm
iv[:5,:3]
```

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(figsize=(10, 6))
ax = fig.add_subplot(projection="3d")
surf = ax.plot_surface(
    strike,
    ttm,
    iv,
    rstride=2,   
    cstride=2,
    cmap=plt.cm.coolwarm,
    linewidth=0.5,
    antialiased=True
)

ax.set_xlabel("strike", labelpad=10)
ax.set_ylabel("time to maturity", labelpad=10)
ax.set_zlabel("implied volatility", labelpad=10)
ax.set_title("Implied Volatility Surface", pad=15)
fig.colorbar(surf, ax=ax, shrink=0.6, aspect=10)

plt.tight_layout()
plt.show()
```
